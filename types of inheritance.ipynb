```python
# 1 . single inheritance
# with implementation in child class
# when child class dont have constructor , in this case parent class constructor behaves like child class constructor 
#parent class
class Person:
    def __init__(self,name,age):
        self.name=name
        self.age=age
    def details(self):
        return f"My name is {self.name} and I an a {self.age} years old"
#child class

class Student(Person):
    pass
s= Student("ravi",20)
print(s.name)
print(s.details())

```

```python
# 1 . single inheritance
# with implementation in child class
# when child class dont have constructor , in this case parent class constructor behaves like child class constructor 
#parent class
class Person:
    def __init__(self,name,age,gender):
        self.name=name
        self.age=age
        self.gender= gender
    def details(self):
        return f"My name is {self.name} and Iam a {self.age} years old {self.gender}"
#child class

class Student(Person):
    def __init__(self ,stdid, name, age):
        self.stdid= stdid
        self.name= name
        self.age= age
P=Person("sakshith",21,"Male")
s= Student(101,"sakshith",20)
print(s.name)
print(P.details())
s.gender = "Male"
print(s.details())
```

```python
# inherete the parent class cosnstructor in child class by using [ super() ]
#super(): This function is user
#single inherita
class Person:
    def __init__(self,name,age,gender):
        self.name=name
        self.age=age
        self.gender= gender
    def details(self):
        return f"My name is {self.name} and Iam a {self.age} years old {self.gender}"
#child class

class Student(Person):
    def __init__(self ,stdid, name, age,gender):
        self.stdid= stdid
        super().__init__(name,age,gender)

        
    def percentage(self,marks:list):
        return sum(marks)/len(marks)

s=Student(101,"sakshith",22,"male")
print(s.name)
print(s.details())
print(s.percentage([90,89,88]))
```

```python
#Multilevel Inheritance.

class Employee:
    company="Sakshithrao It Solutions"
    def __init__(self, id:int,name:str,salary:int):
        self.id= id
        self.name=name
        self.salary=salary
    def calculate_salary(self,bonous=0):
        return self.salary+bonous

        
class Teamlead(Employee):
    def __init__(self,id,name:str,salary:int,project_name:str):
        super().__init__(id,name,salary)
        self.project_name = project_name
    def project_status(self,status):
        return f"{self.project_name}  project status :  {status}% completed"

class Manager(Teamlead):
    def __init__(self,id,name:str,salary:int,project_name:str,no_projects:str):
        super().__init__(id,name,salary,project_name)
        self.no_projects = no_projects
e = Employee(101, "sakshith",35000)
t = Teamlead(102, "rao",45000, "practice tracking project")
m= Manager(103, "sakshith", 55000, " Exam portal project", 5)
print(e.name, t.name , m.name)
print(e.calculate_salary())
print(t.calculate_salary(5000))
print(t.project_status(90))
print(m.calculate_salary(10000))

        
```

```python
# 3 multiple inheritance
class Father:
    def __init__(self, father_name):
        self.father_name = father_name

    def working1(self):
        return "Father working as an Engineer"


class Mother:
    def __init__(self, mother_name):
        self.mother_name = mother_name

    def working2(self):
        return "Mother working as a Teacher"


class Child(Father, Mother):
    def __init__(self, name, father_name, mother_name):
        self.name = name
        Father.__init__(self, father_name)
        Mother.__init__(self, mother_name)

    def working(self):
        print(super().working1())
        print(super().working2())


# Object creation
c = Child("Sakshith", "Satya", "Mahi")

print(c.name)
print(c.father_name)
print(c.mother_name)

c.working()

print(Child.__mro__)
```
