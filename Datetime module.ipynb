```python
# timer 

from datetime import datetime
import time

start_time = datetime.now()

print("Timer started...")
print(start_time)

time.sleep(10)

end_time = datetime.now()

print("Timer stopped...")
print(end_time)

difference = end_time - start_time

print("Time taken:", difference)
```

```python
#stop watch
import time

hours = int(input("Enter hours: "))
minutes = int(input("Enter minutes: "))
seconds = int(input("Enter seconds: "))

total_seconds = hours * 3600 + minutes * 60 + seconds

while total_seconds > 0:
    hours = total_seconds // 3600
    minutes = (total_seconds % 3600) // 60
    seconds = total_seconds % 60

    print(f"{hours:02d}:{minutes:02d}:{seconds:02d}", end="\r")

    time.sleep(1)
    total_seconds -= 1

print("00:00:00")
print("Time's up!")
```

```python
import time

days = 0
hours = 0
minutes = 0
seconds = 0

while True:

    print(f"Day: {days}  {hours:02d}:{minutes:02d}:{seconds:02d}", end="\r")

    time.sleep(1)

    seconds += 1

    if seconds == 60:
        seconds = 0
        minutes += 1

    if minutes == 60:
        minutes = 0
        hours += 1

    if hours == 24:
        hours = 0
        days += 1
```
