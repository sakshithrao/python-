```python
# method overriding
class Person:
    def __init__(self, name, age):
        self.name = name
        self.age = age

    def details(self):
        return f"My name is {self.name} and I am {self.age} years old."


class Student(Person):
    def __init__(self, name, age, rollno):
        super().__init__(name, age)
        self.rollno = rollno

    def details(self):
        return super().details() + " I am a student."


s = Student("Sakshith", 21, 101)

print(s.details())
```

```python
# Method Overloading using default arguments

class Math:

    def add(self, a, b=0, c=0, d=0):
        return a + b + c + d

m = Math()

print(m.add(1, 2, 3, 4))
print(m.add(1, 2, 3))
print(m.add(1, 2))
print(m.add(1))
```

```python
# 3 Operator overloading

class Point:
    def __init__(self, x):
        self.x = x
       
    def __str__(self): # string magic method
        return f" This point class and value is :{self.x}"
    def __add__(self, other):
        return self.x + other.x
    def __sub__(self,other):
        return self.x - other.x
p1 =  Point(5)
p2 =  Point(3)
print(p1.x+p2.x)
print(p1 + p2)
print(p1 - p2)
```
