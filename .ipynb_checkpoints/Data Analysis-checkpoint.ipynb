```python
# numpy array operations 

import numpy as np
arr1 = np.array([1, 2, 3, 4])

print(arr1)
print(np.ndim(arr1))# it returnd dimention of an array
print(np.size(arr1))# it returns total elements inside an array
print(np.shape(arr1))# it returns array  shape
```

```python

arr1 = np.array([1,2,3,4])
arr2 = np.array([4,3,2,1])
print(arr1 - arr2)
print(arr1 + arr2)
print(np. shape(arr1), np.shape(arr2))
print(arr1 +5 )
print(arr1 * 2)
```

```python
import pandas as pd
df = pd.read_csv("Housing.csv")
df
```

```python
# Check the number of rows and columns
print("Shape:", df.shape)

# Check the number of dimensions
print("Dimensions:", df.ndim)

# Check total number of values
print("Size:", df.size)

# Display column names
print("Columns:")
print(df.columns)

df.head()
df.tail()
df.info()


df.isnull().sum()


df.duplicated().sum()


df.describe()

print("Average price:", df["price"].mean())
print("Median price:", df["price"].median())
print("Minimum price:", df["price"].min())
print("Maximum price:", df["price"].max())


df["airconditioning"].value_counts()


df["furnishingstatus"].value_counts()

df.corr(numeric_only=True)

df["price"].corr(df["area"])

import matplotlib.pyplot as plt

plt.scatter(df["area"], df["price"])

plt.xlabel("Area")
plt.ylabel("Price")
plt.title("Area vs Price")

plt.show()



plt.boxplot(df["price"])

plt.ylabel("Price")
plt.title("House Price Outliers")

plt.show()


print(df.shape)

print(df.columns)

df.info()

print(df.isnull().sum())

print(df.duplicated().sum())
```
