```python
# random module

import random
print(random.random()) #this function returns random balues in between 0 to 1 
print(random.randint(1,10))# it genrate random number in between a and b 
print(random.choice(['A',1,2,3,4,5,6,7,8,9,'j','Q','k'])) # it returns random items from sequence given 
print(random.choices(['A',1,2,3,4,5,6,7,8,9,'j','Q','k'], k = 3))
arr=(['A',1,2,3,4,5,6,7,8,9,'j','Q','k'])
random.shuffle(arr)
print(arr)
```

```python
import random
steps = rolls =0
while True:
    input("To roll dice for  that click enter button")
    dice_val = random.randint(1,6)
    rolls +=1
    print("Current dice value:", dice_val)
    if steps + dice_val <=100:
        steps += dice_val
    if steps == 100:
        print(f"your reached TOP in {rolls} rolls")
        break
        
    print(f"For {rolls} rolls your at {steps} position")
```
