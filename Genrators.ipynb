```python
# genrators defination

def gen(x):
    x=x+1
    yield x
    x=x+2
    yield x
    x=x+3
    yield x
res=gen(5)
print(res)
print(next(res))
print(next(res))
print(next(res))
#print(next(res)) # if it doesn not contain any value it returns  stop iteration error ( to over come this error we use loops
```

```python
# to overcome the stop iteration error we use loops
def gen(x):
    x=x+1
    yield x
    x=x+2
    yield x
    x=x+3
    yield x
res=gen(5)
for i in res:
    print(i)
    
```

```python
# using genral function 
def fun(x):
    res=[]
    for i in range(1,x+1):
        res .append(i)
    return res
print(fun(10))

```
