```python
age = int(input("Enter your age: "))
health_score = int(input("Enter your health score (1-100): "))
vehicle_type = input("Enter your vehicle type (Sports_Car/SUV/Sedan): ").lower()

base_premium = 10000

# Age factor
if age < 25:
    premium = base_premium * 1.5
elif age <= 50:
    premium = base_premium
else:
    premium = base_premium * 1.2

# Health factor
if health_score >= 80:
    premium = premium * 0.8
elif health_score >= 50:
    premium = premium
else:
    premium = premium * 1.2

# Vehicle factor
if vehicle_type == "sports_car":
    premium = premium * 1.3
elif vehicle_type == "suv":
    premium = premium * 1.5
elif vehicle_type == "sedan":
    premium = premium
else:
    premium = premium * 1.1

print("Final Insurance Premium:", premium)
```
