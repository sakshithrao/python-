```python
def factorial(n):
    if n==1 or n==0:
        return 1
    return n*factorial(n-1)
factorial(4)
```

```python
def sum_natural(n):
    if n == 1:          # Base case
        return 1

    return n + sum_natural(n - 1)   # Recursive case

num = 5
print("Sum =", sum_natural(num))
```

```python
# find square of all list of integers using lambda and map()

arr=[1,2,3,4,5]
res=list(map(lambda x : x ** 2,arr))
print(res)
```

```python
# find even numbers in list using lambda and filter()

arr=[1,2,3,4,5]
res=list(filter(lambda x : x%2==0, arr))
print(res)
```
